```python
import matplotlib.pyplot as plt
import numpy as np
import math
```

```python
def b(k):
    beta = 0.4
    return (math.exp(beta)/(math.exp(beta)-1))*(math.exp(beta*k)-1)

def g(eps,beta,z):
    r = 3
    return math.log(math.exp(beta*r)+1)/((math.exp(eps)-1)*math.log(z**r+1)+math.log(math.exp(beta*r)+1))

def h(z):
    eps = 0.5; beta = 0.4
    if math.fabs(z) < math.exp(beta):
        k = 0
    elif math.fabs(z) < math.exp(beta)+math.exp(2*beta):
        k = 1
    elif math.fabs(z) < math.exp(beta)+math.exp(2*beta)+math.exp(3*beta):
        k = 2
    elif math.fabs(z) < math.exp(beta)+math.exp(2*beta)+math.exp(3*beta)+math.exp(4*beta):
        k = 3
    else:
        k = 4
    return (1/math.exp(eps*k))*g(eps,beta,(math.fabs(z)-b(k))/math.exp(beta*k))

fig = plt.figure()
z1 = np.linspace(-12,0,1000); z2 = np.linspace(0,12,1000)
plt.plot(z1,[h(z1[i]) for i in range(1000)],color="k")
plt.plot(z2,[h(z2[i]) for i in range(1000)],color="k")
plt.ylim(0,1.05)
plt.xlim(-9.5,9.5)
plt.xticks([-b(3),-b(2),-b(1),0,b(1),b(2),b(3)],["$-b_3$", "$-b_2$", "$-b_1$", "$b_0$", "$b_1$", "$b_2$", "$b_3$"],fontsize=14)
plt.yticks([h(b(3)),h(b(2)),h(b(1)),h(0)], [r"$\frac{a}{e^{3\epsilon'}}$",r"$\frac{a}{e^{2\epsilon'}}$",r"$\frac{a}{e^{\epsilon'}}$",r"$a$"],fontsize = 17)
plt.plot([-9.5,0], [h(0),h(0)], color="black", linestyle='dashed', lw=0.6)
plt.plot([-9.5,b(1)], [h(b(1)),h(b(1))], color="black", linestyle='dashed', lw=0.6)
plt.plot([-9.5,b(2)], [h(b(2)),h(b(2))], color="black", linestyle='dashed', lw=0.6)
plt.plot([-9.5,b(3)], [h(b(3)),h(b(3))], color="black", linestyle='dashed', lw=0.6)
plt.plot([0,0], [0,h(0)], color="black", linestyle='dashed', lw=0.6)
plt.plot([b(1),b(1)], [0,h(b(1))], color="black", linestyle='dashed', lw=0.6)
plt.plot([b(2),b(2)], [0,h(b(2))], color="black", linestyle='dashed', lw=0.6)
plt.plot([b(3),b(3)], [0,h(b(3))], color="black", linestyle='dashed', lw=0.6)
plt.plot([-b(1),-b(1)], [0,h(b(1))], color="black", linestyle='dashed', lw=0.6)
plt.plot([-b(2),-b(2)], [0,h(b(2))], color="black", linestyle='dashed', lw=0.6)
plt.plot([-b(3),-b(3)], [0,h(b(3))], color="black", linestyle='dashed', lw=0.6)
plt.xlabel("$z$", fontsize = 18)
plt.ylabel("$h(z)$", fontsize = 18)
fig.savefig("SID-V.png")
fig.savefig("figs/SID-V.eps", bbox_inches="tight", pad_inches=0.05)
```
