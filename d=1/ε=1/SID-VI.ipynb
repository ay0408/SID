```python
from sympy import *
from scipy.optimize import minimize
from scipy import integrate
from scipy.integrate import quad
```

```python
def E(peps,beta):
    return 1/(exp(peps-2*beta)-1) + ((exp(peps-beta)-1)/(exp(peps-2*beta)-1))*(1/2)

def a(peps,beta):
    return peps
```

```python
eps = 1; beta = 0

fun = lambda x: E(x[0],beta) 
bnds = ((2*beta,eps), (0,None))
cons = ({'type': 'ineq', 'fun': lambda x:  eps-beta-a(x[0],beta)},)
res = minimize(fun, (eps,0), bounds = bnds, constraints = cons)
print(res.fun, res.x)
```
