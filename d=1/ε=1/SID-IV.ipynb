```python
from sympy import *
from scipy.optimize import minimize
from scipy import integrate
from scipy.integrate import quad
```

```python
def E(peps,beta,r):
    c,e = quad(lambda z: z*(-((exp(peps)-1)/(exp(peps)*log(exp(beta*r)+1)))*log(z**r+1)+1),0,exp(beta))
    m,e = quad(lambda z: -((exp(peps)-1)/(exp(peps)*log(exp(beta*r)+1)))*log(z**r+1)+1,0,exp(beta))
    return 1/(exp(peps-2*beta)-1) + ((exp(peps-beta)-1)/(exp(peps-2*beta)-1))*(1/exp(beta))*(c/m)

def g(z,beta,peps,r):
    return -((exp(peps)-1)/(exp(peps)*log(exp(beta*r)+1)))*log(z**r+1)+1

def S(peps,beta,r):
    fun1 = lambda z: -g(z[0],beta,peps,r)/g(z[0]+1,beta,peps,r)
    bnds1 = ((0,exp(beta)-1), (0,None))
    res1 = minimize(fun1, (0,0), bounds = bnds1)
    #print(log(-res1.fun).evalf())
    fun2 = lambda z: -exp(peps)*g(z[0],beta,peps,r)/g((z[0]+1-exp(beta))/exp(beta),beta,peps,r)
    bnds2 = ((exp(beta)-0.999,exp(beta)), (0,None))
    res2 = minimize(fun2, (exp(beta)-1, 0), bounds = bnds2)
    #print(log(-res2.fun).evalf())
    x = max(-res1.fun, -res2.fun)
    return log(x)
```

```python
eps = 1; beta = 0
#peps = x[0], r = x[1]

fun = lambda x: E(x[0],beta,x[1]) 
bnds = ((2*beta,eps), (0,None))
cons = ({'type': 'ineq', 'fun': lambda x:  eps-beta-S(x[0],beta,x[1])},)
res = minimize(fun, (2*beta+1,1), bounds = bnds, constraints = cons)
print(res.fun.evalf(), res.x)
```
