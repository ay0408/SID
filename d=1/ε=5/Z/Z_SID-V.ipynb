```python
from sympy import *
from scipy.optimize import minimize
from scipy import integrate
from scipy.integrate import quad
import matplotlib.pyplot as plt
import numpy as np
import math
import random
```

```python
#beta = 0
def g(z):
    return math.log(2)/((math.exp(5)-1)*math.log(z**1.9806237+1)+math.log(2))

def b(k):
    return k

def S(k):
    return (1/2)*(1-(1/math.exp(5*k)))

def ComputeZ(d):
    c = random.uniform(-0.5,0.5)
    #print(c)
    k = 0
    while(1):
        if S(k) <= math.fabs(c) and math.fabs(c) < S(k+1):
            break
        else:
            k += 1
    a = ((math.exp(5)-1)/(2*math.exp(5)))*(1/quad(lambda z: g(z),0,1)[0])
    fun = lambda x: math.fabs(math.fabs(c)-S(k)-(1/exp(5*k))*quad(lambda z: a*g(z),0,x[0])[0])
    bnds = ((0,1), (0,None))
    res = minimize(fun, (0.5,1), bounds = bnds)
    #print(res.fun, res.x)
    return (b(k)+res.x[0])*(c/math.fabs(c))            
```

```python
def Evaluation(d):
    Z = 0
    for i in range(100):
        Z += math.fabs(ComputeZ(d))
    return Z/100
```

```python
ave = 0; M = 0; m = 10000000
for i in range(100):
    Z = Evaluation(1)
    M = max(Z,M); m = min(Z,m)
    print(Z)
    ave += Z
    
print("-----")
print(ave/100)
print(M, m)
```

```python
beta = 0.4; peps = 5; r = 2.13620418
def g(z):
    return math.log(math.exp(beta*r)+1)/((math.exp(peps)-1)*math.log(z**r+1)+math.log(math.exp(beta*r)+1))

def b(k):
    return (math.exp(beta)/(math.exp(beta)-1))*(math.exp(beta*k)-1)

def S(k):
    return (1/2)*(1-(1/math.exp((peps-beta)*k)))

def ComputeZ(d):
    c = random.uniform(-0.5,0.5)
    #print(c)
    k = 0
    while(1):
        if S(k) <= math.fabs(c) and math.fabs(c) < S(k+1):
            break
        else:
            k += 1
    a = ((math.exp(peps-beta)-1)/(2*math.exp(peps-beta)))*(1/quad(lambda z: g(z),0,math.exp(beta))[0])
    fun = lambda x: math.fabs(math.fabs(c)-S(k)-(1/exp((peps-beta)*k))*quad(lambda z: a*g(z),0,x[0]/math.exp(beta*k))[0])
    bnds = ((0,math.exp(beta*(k+1))), (0,None))
    res = minimize(fun, (math.exp(beta*k),1), bounds = bnds)
    #print(res.fun, res.x)
    return (b(k)+res.x[0])*(c/math.fabs(c))            
```

```python
ave = 0; M = 0; m = 10000000
for i in range(100):
    Z = Evaluation(1)
    M = max(Z,M); m = min(Z,m)
    print(Z)
    ave += Z
    
print("-----")
print(ave/100)
print(M, m)
```
