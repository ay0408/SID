```python
from scipy.optimize import minimize
from scipy import integrate
from scipy.integrate import quad
import matplotlib.pyplot as plt
import numpy as np
import math
import random
```

```python
def h(z):
    return (5/2)*math.exp(-5*math.fabs(z))

def ComputeZ(d):
    c = random.uniform(-0.5,0.5)
    fun = lambda x: math.fabs(math.fabs(c)-quad(lambda z: h(z), 0, x[0])[0])
    bnds = ((0,None), (0,None))
    res = minimize(fun, (1,1), bounds = bnds)
    return res.x[0]*(c/math.fabs(c))
```

```python
def Evaluation(d):
    Z = 0
    for i in range(100):
        Z += math.fabs(ComputeZ(d))
    return Z/100
```

```python
ave = 0; M = 0; m = 10000000
for i in range(100):
    Z = Evaluation(1)
    M = max(Z,M); m = min(Z,m)
    print(Z)
    ave += Z
    
print("-----")
print(ave/100)
print(M, m)
```
