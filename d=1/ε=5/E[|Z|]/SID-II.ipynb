```python
from sympy import *
from scipy.optimize import minimize
from scipy import integrate
from scipy.integrate import quad
```

```python
def E(peps,beta,r):
    return 1/(exp(peps-2*beta)-1) + ((exp(peps-beta)-1)*(r+1)*(r*exp(peps)+2))/(2*(exp(peps-2*beta)-1)*(r+2)*(r*exp(peps)+1))

def S(peps,beta,r):
    return peps+log(1-((exp(peps)-1)/exp(peps+r*beta))*((exp(beta)-1)**r))
```

```python
eps = 5; beta = 0
#peps = x[0], r = x[1]

fun = lambda x: E(x[0],beta,x[1]) 
bnds = ((2*beta,eps), (0,None))
cons = ({'type': 'ineq', 'fun': lambda x:  eps-beta-S(x[0],beta,x[1])},)
res = minimize(fun, (2*beta+1,1), bounds = bnds, constraints = cons)
print(res.fun.evalf(), res.x)
```
