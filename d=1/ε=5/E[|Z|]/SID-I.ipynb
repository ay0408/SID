```python
from sympy import *
from scipy.optimize import minimize
from scipy import integrate
from scipy.integrate import quad
```

```python
def E(peps,beta,r):
    c,e = quad(lambda z: z*exp(-(peps/exp(beta*r))*(z**r)),0,exp(beta))
    m,e = quad(lambda z: exp(-(peps/exp(beta*r))*(z**r)),0,exp(beta))
    return 1/(exp(peps-2*beta)-1) + ((exp(peps-beta)-1)/(exp(peps-2*beta)-1))*(1/exp(beta))*(c/m)

def S(peps,beta,r):
    return peps*(1-(1-1/exp(beta))**r)
```

```python
eps = 5; beta = 0
#peps = x[0], r = x[1]

fun = lambda x: E(x[0],beta,x[1]) 
bnds = ((2*beta,eps), (0,None))
cons = ({'type': 'ineq', 'fun': lambda x:  eps-beta-S(x[0],beta,x[1])},)
res = minimize(fun, (2*beta+1,1), bounds = bnds, constraints = cons)
print(res.fun.evalf(), res.x)
```
