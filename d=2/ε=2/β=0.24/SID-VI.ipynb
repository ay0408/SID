```python
from sympy import *
from scipy.optimize import minimize
from scipy import integrate
from scipy.integrate import quad
from mpmath import nsum, inf
```

```python
def E(peps,beta): #beta > 0
    m = lambda k: (1/exp(peps*k))*quad(lambda z1: quad(lambda z2: 1, max(0,((exp(beta)/(exp(beta)-1))*(exp(beta*k)-1))-z1), ((exp(beta)/(exp(beta)-1))*(exp(beta*(k+1))-1))-z1)[0], 0, ((exp(beta)/(exp(beta)-1))*(exp(beta*(k+1))-1)))[0]
    M = nsum(lambda k: m(k), [0,inf])
    #print(M)

    c = lambda k: (1/exp(peps*k))*quad(lambda z1: quad(lambda z2: (z1+z2), max(0,((exp(beta)/(exp(beta)-1))*(exp(beta*k)-1))-z1), ((exp(beta)/(exp(beta)-1))*(exp(beta*(k+1))-1))-z1)[0], 0, ((exp(beta)/(exp(beta)-1))*(exp(beta*(k+1))-1)))[0]
    C = nsum(lambda k: c(k), [0,inf])
    #print(C)
    
    return C/M

def E2(peps,beta): #beta = 0
    m = lambda k: (1/exp(peps*k))*quad(lambda z1: quad(lambda z2: 1, max(0,k-z1), k+1-z1)[0], 0, k+1)[0]
    M = nsum(lambda k: m(k), [0,inf])
    #print(M)

    c = lambda k: (1/exp(peps*k))*quad(lambda z1: quad(lambda z2: (z1+z2), max(0,k-z1), k+1-z1)[0], 0, k+1)[0]
    C = nsum(lambda k: c(k), [0,inf])
    #print(C)
    
    return C/M

def S(peps,beta):
    return peps
```

```python
eps = 2; beta = 0.24
#peps = x[0]

fun = lambda x: E(x[0],beta) 
bnds = ((3*beta,eps), (0,None))
cons = ({'type': 'ineq', 'fun': lambda x:  eps-beta*2-S(x[0],beta)},)
res = minimize(fun, (3*beta+0.1,0), bounds = bnds, constraints = cons)
print(res.fun, res.x)
```
