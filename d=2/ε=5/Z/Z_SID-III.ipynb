```python
from sympy import *
from scipy.optimize import minimize
from scipy import integrate
from scipy.integrate import quad
import matplotlib.pyplot as plt
import numpy as np
import math
import random
from mpmath import nsum, inf
```

```python
#beta = 0
def g(z):
    return 1/((math.exp(5)-1)*(abs(z)**3.30769321)+1)

def b(k):
    return k

def X(k):
    return quad(lambda z1: quad(lambda z2: g(z1+z2-b(k)), max(0,b(k)-z1), b(k+1)-z1)[0], 0, b(k+1))[0]

def S(a,k):
    T = nsum(lambda i: (1/math.exp(5*i))*X(i), [0,k-1])
    return a*T

def ComputeZ(a,d):
    c = random.uniform(0,1/(2**d))
    #print(c)
    k = 0
    while(1):
        if S(a,k) <= math.fabs(c) and math.fabs(c) < S(a,k+1):
            break
        else:
            k += 1
    #print(math.fabs(c)-S(a,k))
    fun = lambda x: math.fabs(math.fabs(c)-S(a,k)-a*(1/math.exp(5*k))*quad(lambda z1: quad(lambda z2: g(z1+z2-b(k)), max(0,b(k)-z1), b(k)+x[0]-z1)[0], 0, b(k)+x[0])[0])
    bnds = ((0,1), (0,None))
    res = minimize(fun, (0.1,1), bounds = bnds)
    #print(res.fun, res.x)
    return b(k)+res.x[0] #||Z||_1
```

```python
def Evaluation(a,d):
    Z = 0
    for i in range(100):
        Z += ComputeZ(a,d)
    return Z/100
```

```python
a = 1/(4*nsum(lambda k: (1/math.exp(5*k))*X(k), [0,inf]))
ave = 0; M = 0; m = 10000000
for i in range(100):
    Z = Evaluation(a,2)
    M = max(Z,M); m = min(Z,m)
    print(Z)
    ave += Z
    
print("-----")
print(ave/100)
print(M, m)
```

```python
beta = 0.2; peps = 4.9938006; r = 2.94509805
def g(z):
    return math.exp(beta*r)/((math.exp(peps)-1)*(abs(z)**r)+math.exp(beta*r))

def b(k):
    return (math.exp(beta)/(math.exp(beta)-1))*(math.exp(beta*k)-1)

def X(k):
    return quad(lambda z1: quad(lambda z2: g((z1+z2-b(k))/math.exp(beta*k)), max(0,b(k)-z1), b(k+1)-z1)[0], 0, b(k+1))[0]

def S(a,k):
    T = nsum(lambda i: (1/math.exp(peps*i))*X(i), [0,k-1])
    return a*T

def ComputeZ(a,d):
    c = random.uniform(0,1/(2**d))
    #print(c)
    k = 0
    while(1):
        if S(a,k) <= math.fabs(c) and math.fabs(c) < S(a,k+1):
            break
        else:
            k += 1
    #print(math.fabs(c)-S(a,k))
    fun = lambda x: math.fabs(math.fabs(c)-S(a,k)-a*(1/math.exp(peps*k))*quad(lambda z1: quad(lambda z2: g((z1+z2-b(k))/math.exp(beta*k)), max(0,b(k)-z1), b(k)+x[0]-z1)[0], 0, b(k)+x[0])[0])
    bnds = ((0,math.exp(beta*(k+1))), (0,None))
    res = minimize(fun, (0.1,1), bounds = bnds)
    #print(res.fun, res.x)
    return b(k)+res.x[0] #||Z||_1
```

```python
a = 1/(4*nsum(lambda k: (1/math.exp(peps*k))*X(k), [0,inf]))
ave = 0; M = 0; m = 10000000
for i in range(100):
    Z = Evaluation(a,2)
    M = max(Z,M); m = min(Z,m)
    print(Z)
    ave += Z
    
print("-----")
print(ave/100)
print(M, m)
```
