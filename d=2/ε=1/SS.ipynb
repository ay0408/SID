```python
import numpy as np
import math
import matplotlib.pyplot as plt
```

```python
def SS1(eps,beta):
    k = 2-(6*beta*2)/eps
    alpha = (eps/(2*(3**(3/4))))*k
    return (1/(alpha*(2**(1/2))))*2

def SS2(eps,beta):
    k = 2-(6*beta*2)/eps
    alpha = (eps/4)*k
    return ((1/alpha)*(2/(math.pi)))*2

def SS3(eps,beta):
    k = 2-(10*beta*2)/eps
    alpha = (eps/(2*(5**(5/6))))*k
    return (1/(alpha*(3**(1/2))))*2
```

```python
beta = [0.04*i for i in range(9)]

eps = 1
for i in range(5):
    print(SS1(eps,beta[i]))
```

```python
for i in range(5):
    print(SS2(eps,beta[i]))
```

```python
for i in range(3):
    print(SS3(eps,beta[i]))
```
